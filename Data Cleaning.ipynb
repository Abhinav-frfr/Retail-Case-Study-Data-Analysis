```python
df['tran_date'] = pd.to_datetime(df['tran_date'], dayfirst=True,format='mixed')

df['Transaction_Date'] = df['tran_date'].dt.day
df['Transaction_Month'] = df['tran_date'].dt.month
df['Transaction_Year'] = df['tran_date'].dt.year
```

```python
def encode_Stores(store:str) -> int:
    if store=='e-Shop':
        return 1
    elif store=='MBR':
        return 2
    elif store=='Flagship store':
        return 3
    elif store=='TeleShop':
        return 4  
```

```python
df['Gender_encoded']=df['Gender'].apply(lambda x: 1 if x=='M' else 0)
df['Store_type_encoded']=df['Store_type'].apply(encode_Stores)
```

```python
df.head()
```

```python
df.drop(columns=['Gender','Store_type'],inplace=True)
```

```python
df['Store_type_encoded']=df['Store_type_encoded'].astype(int)
```
