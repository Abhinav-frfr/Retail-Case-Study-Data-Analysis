```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Retail_Case_Study_Cleaned.csv')
```

```python
df.head()
```

```python
df['prod_cat_code'].value_counts()
```

```python
sns.countplot(data=df,x='prod_cat_code')

## Looks Like Product 5 is Bought highly while Product 4 is hardly bought. 
```

```python
sns.countplot(data=df,x='prod_subcat_code',color='green')

## Product with Sub-Category 4 is purchased a lot while product with Sub-Category 2,5,6,7,8,9 are purchased very less.
```

```python
df2=pd.DataFrame(df.groupby(['prod_cat_code','prod_subcat_code'])['Rate'].mean())
df2
```

```python
sns.scatterplot(data=df2,x='prod_cat_code',y='prod_subcat_code',hue='Rate')

## Product 5 with sub category 10,11,12 and Product 4 with sub category 1 and Product 3 with sub category 5 and Product 1 with sub category 3 are expensive
```

```python
df3=pd.DataFrame(df.groupby(['prod_cat_code','prod_subcat_code'])['Tax'].mean())
df3
```

```python
sns.scatterplot(data=df3,x='prod_cat_code',y='prod_subcat_code',hue='Tax')

## The most tax is paid over Product 5 (Sub-Category 12), Product 3 (Sub-Category 4,5), Product 2 (Sub-Category 4), Product 1 (Sub-Category 1,3,4)
```

```python
df.head()
```

```python
df4=pd.DataFrame(df.groupby(['prod_cat_code','prod_subcat_code'])['Qty'].agg(['mean','max','sum']))
df4
```

```python
sns.scatterplot(data=df4,x='prod_cat_code',y='prod_subcat_code',hue='sum')

## Product 3 (Sub-Cat 4), Product 5 (Sub-Cat 7), Product 6 (Sub-Cat 12) are purchased a lot
# Meaning they are good products or more of a general requirement 
```

```python
df.head()
```

```python
df4=pd.DataFrame(df.groupby(['Transaction_Year'])['total_amt'].agg(['mean','sum','median']))
df4.head()
```

```python
df5=df[['Transaction_Year']]
df5['Counter']=1
df6=pd.DataFrame(df5.groupby(['Transaction_Year'])['Counter'].agg(['sum']))
df6.head()

## Most amount of sale was done in year 2012 and then dropped over the subsequent 2 years. However, mean and median sale is highest in year 2011.
```

```python
df['customer_Id'].value_counts()
```

```python
df7=pd.DataFrame(df.groupby(['customer_Id'])['total_amt'].agg(['mean','sum','max']))
df7
```

```python
df7[df7['sum']>40000]

## These are high value Customers who have purchased products worth more than 40000 in span of 4 years. There are 5 such customers.
```

```python
df8=pd.DataFrame(df.groupby(['Transaction_Month'])['total_amt'].agg(['mean','sum','median']))
df8

## Store appears to be an All round Season Store as products are equally sold though the year. However, the highest sale is in month of February and lowest in month of June.
```

```python
df.head()
```

```python
df['Gender_encoded'].value_counts()

## The store has almost equal number of male and female customers. However, the number of female customers is slightly higher than male customers.
```

```python
df.groupby(['Gender_encoded'])['total_amt'].agg(['mean','sum','median'])

## Higher total amount purchased by female customers. However, mean and median amount purchased is higher for male customers. This indicates that though female customers are more in number, male customers are high value customers.
```

```python
sns.scatterplot(data=df,x='city_code',y='total_amt',hue='Gender_encoded',legend='auto')
```

```python
df.groupby(['city_code'])['total_amt'].agg(['mean','sum','median'])


## City pattern shows that city 3 has the highest total amount purchased. City 6 has the lowest total amount purchased.
```

```python
df.groupby(['city_code','Gender_encoded'])['total_amt'].agg(['mean','sum','median'])

# Gender ratio Purchasing is different in different cities. City 3 has more female customers while city 1 has more male customers.
```

```python
df5=df[['city_code','prod_cat_code','prod_subcat_code','total_amt']]
df5['Counter']=1
df6=pd.DataFrame(df5.groupby(['city_code','prod_cat_code'])['Counter'].agg(['sum']))
df6

## For each City the different product category purchased is different which can be observed here.
```

```python
ax = sns.scatterplot(data=df6,x='city_code',y='prod_cat_code',hue='sum')

# Remove the legend object from the axes
ax.legend_.remove()

plt.show()

## It appears Product 5 is prevelant in every city while Product 4 is hardly purchased in any city.
```
